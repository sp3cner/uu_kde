```python
#q3

#a
X = np.array([5, 7, 7, 8, 10])
h = 1.5
n = len(X)

for x in [6.0, 6.5, 7.0, 8.0, 8.5]:
    count = np.sum(np.abs(x - X) <= h)
    print(x, count, count / (2 * n * h))
```

```python
#b
for x in [6.0, 6.5, 7.0, 8.0, 8.5]:
    u = (x - X) / h
    print(x, norm.pdf(u).sum() / (n * h))
```

```python
#q4
df = pd.read_csv("../data/heart_failure_clinical_records_dataset.csv")
df.head()

#a
sns.kdeplot(data=df, x="ejection_fraction")
plt.xlabel("Ejection fraction (%)")
plt.title("KDE of ejection fraction")
plt.show()

#b
for i in range(15):
    boot = df.sample(frac=1.0, replace=True, random_state=i)
    sns.kdeplot(data=boot, x="ejection_fraction", color="gray", alpha=0.4, linewidth=1)

sns.kdeplot(data=df, x="ejection_fraction", color="black", linewidth=2, label="Original")
plt.xlabel("Ejection fraction (%)")
plt.title("Original KDE vs. 15 bootstrap resamples")
plt.legend()
plt.show()
```
